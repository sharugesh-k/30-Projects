```python
director_list=pd.DataFrame()
```

```python
director_list=df['director'].str.split(',',expand=True).stack()
print(director_list)
```

```python
director_list=director_list.to_frame()
```

```python
print(director_list)
```

```python
director_list.columns=['director']
print(director_list)
```

```python
directors=director_list.groupby(['director']).size().reset_index(name='counts')
print(directors)
```

```python
directors=directors[directors.director !='not specified']
```

```python
directors=directors.sort_values(by=['counts'],ascending=False)
print(directors)
```
